# Part A: Vector & Matrix Fundamentals

```python
import numpy as np
import pandas as pd

data = pd.read_csv("Dataset.csv")
data.head()
```

## 1. Represent each student’s subject scores as a vector.

```python
vectors = {}
student_id = data['Student_ID']
scores = data[['Math', 'Science', 'English']]
vectors = scores
print(student_id, scores)
```

## 2. Compute:

### i) Norm-1 and Norm-2 of vectors.

```python
vector = np.array(scores, dtype=float)
l1_norms = np.linalg.norm(vector, ord=1)
l2_norms = np.linalg.norm(vector, ord=2)

print("L1 norm of the vector:", l1_norms)
print("L2 norm of the vector:", l2_norms)
```

### ii) Dot product and angle between two students’ score vectors.

```python
vector1 = np.array([85, 90, 80, 95, 75], dtype=float)
vector2 = np.array([92, 95, 89, 96, 85], dtype=float)
dot_product = np.dot(vector1, vector2)
print("Dot product of the vectors:", dot_product)
angle = np.arccos(dot_product / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle between the vectors:", angle)
```

### iii) Cross product (for 3D selected subjects)

```python
vectors1 = np.array([85, 90, 80], dtype=float)
vectors2 = np.array([92, 95, 89], dtype=float)
cross_product = np.cross(vectors1, vectors2)
print("Cross product of the vectors:", cross_product)
```

## 3. Find the projection of one vector onto another.

```python
projection = (np.dot(vector1, vector2) / np.linalg.norm(vector2)**2) * vector2
print("Projection of vector1 onto vector2:", projection)
```

# Part C: Linear Transformations & Geometry

## 5. Explain line, plane, and hyperplane with respect to your dataset dimensions.

### (a) Line (1D)

A **line** is one-dimensional geometric object that extends infinitely in two directions.

**Equation:**

\[
y = mx + c
\]

**Dataset Example:**

If only **Math** marks are considered, each student is represented by a single value.

Example:

| Student | Math |
|---------|------|
| Amit | 85 |
| Riya | 78 |
| Neha | 92 |

These values lie along one dimension (a line).

---

### (b) Plane (2D)

A **plane** is a flat two-dimensional surface.

**Equation:**

\[
ax + by + c = 0
\]

**Dataset Example:**

Using two subjects:

- Math
- Science

Each student is represented as a point:

| Student | Coordinates |
|---------|-------------|
| Amit | (85, 90) |
| Riya | (78, 88) |
| Neha | (92, 95) |

These points lie on a **2-dimensional plane**.

---

### (c) Hyperplane (Higher Dimensions)

A **hyperplane** is a flat surface that exists in more than three dimensions.

**Equation:**

\[
a_1x_1+a_2x_2+\cdots+a_nx_n+b=0
\]

**Dataset Example:**

Using all five subjects:

- Math
- Science
- English
- Computer
- History

Each student becomes a point in **5-dimensional space**.

Example:

Amit

\[
(85,\;90,\;80,\;95,\;75)
\]

Machine learning algorithms often use hyperplanes to separate different classes of data.

---

# Part D: Eigenvalues & Decomposition

```python
import numpy as np
import pandas as pd
from scipy.linalg import lu
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
import matplotlib.pyplot as plt

students = pd.DataFrame({
    "Student": ["Amit","Riya","Neha","Rahul","Priya"],
    "Math": [85,78,92,70,88],
    "Science": [90,88,95,75,91],
    "English": [80,84,89,72,86],
    "Computer": [95,91,96,78,93],
    "History": [75,82,85,74,84]
})

```

## 7. Compute the eigenvalues and eigenvectors of the covariance matrix

```python
# 7. Compute the eigenvalues and eigenvectors of the covariance matrix

X = students.iloc[:,1:].values

print(X)

cov_matrix = np.cov(X.T)

print("Covariance Matrix:\n")
print(cov_matrix)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("\nEigenvalues\n")
print(eigenvalues)

print("\nEigenvectors\n")
print(eigenvectors)
```

## 8. Perform LU Decomposition of the dataset matrix.

```python
# 8. Perform LU Decomposition of the dataset matrix.

P, L, U = lu(X)
print("LU Decomposition:")
print("P:\n", P)
print("L:\n", L)
print("U:\n", U)
```

## 9. Perform Singular Value Decomposition (SVD) and explain its role in dimensionality reduction

```python
# 9. Perform Singular Value Decomposition (SVD) and explain its role in dimensionality reduction.

U, S, Vt = np.linalg.svd(X)
print("Singular Value Decomposition:")
print("U:\n", U)
print("S:\n", S)
print("V transpose:\n", Vt)
```

### Role in Dimensionality Reduction

- SVD identifies the most important patterns in the data.
- Larger singular values capture most of the information.
- Smaller singular values usually represent noise or less important information.
- By keeping only the largest singular values, the dimensionality of the dataset can be reduced while preserving most of the original information.
- SVD is widely used in machine learning, recommendation systems, image compression, natural language processing, and Principal Component Analysis (PCA).

# Part-E: Dimensionality Reduction

## 10. Apply Principal Component Analysis (PCA) to reduce the dataset from multiple subjects to 2 dimensions.

```python
# 10. Apply Principal Component Analysis (PCA) to reduce the dataset from multiple subjects to 2 dimensions.

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load Student Dataset
students = pd.read_csv("Dataset.csv")

# Select only subject columns
X = students[["Math", "Science", "English", "Computer", "History"]]

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA (Reduce from 5 dimensions to 2)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print("PCA Transformed Data:")
print(X_pca)

# Create DataFrame for PCA result
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Student"] = students["Student_Name"]

# Print explained variance
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Print transformed data
print("\nPCA Transformed Data:")
print(pca_df)

# Plot
plt.figure(figsize=(8,6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    color="royalblue",
    s=80,
    edgecolors="black"
)

# Display student names
for i in range(len(pca_df)):
    plt.annotate(
        pca_df["Student"][i],
        (pca_df["PC1"][i], pca_df["PC2"][i]),
        xytext=(5,5),
        textcoords="offset points"
    )

plt.title("PCA of Student Subject Scores")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)

plt.show()
```

## 11. Apply Linear Discriminant Analysis (LDA) to classify students into “Above Average” and “Below Average” categories.

```python
# 11. Apply Linear Discriminant Analysis (LDA) to classify students into “Above Average” and “Below Average” categories.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# Load Student Dataset
students = pd.read_csv("Dataset.csv")

# Select subject columns
X = students[["Math", "Science", "English", "Computer", "History"]]

# Create class labels based on average marks
students["Average"] = X.mean(axis=1)

students["Category"] = np.where(
    students["Average"] >= 85,
    "Above Average",
    "Below Average"
)

# Convert categories into numeric labels
y = np.where(students["Category"] == "Above Average", 1, 0)

# Apply LDA
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

# Create result DataFrame
lda_df = pd.DataFrame({
    "Student": students["Student_Name"],
    "LDA1": X_lda.flatten(),
    "Category": students["Category"]
})

print(lda_df)
```

```python
# Visualize the LDA result
plt.figure(figsize=(8,6))

plt.scatter(
    lda_df["LDA1"],
    np.zeros_like(lda_df["LDA1"]),
    c=np.where(lda_df["Category"] == "Above Average", "green", "red"),
    s=80,
    edgecolors="black"
)

for i in range(len(lda_df)):
    plt.annotate(
        lda_df["Student"][i],
        (lda_df["LDA1"][i], 0),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Linear Discriminant Analysis (LDA) - Student Dataset")
plt.xlabel("Linear Discriminant 1")
plt.yticks([])
plt.grid(True)

plt.show()
```
