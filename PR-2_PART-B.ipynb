# Part B – Data Analysis & Testing Tasks

## 1.Formulate at least two hypotheses

```python
import pandas as pd
import numpy as np
df = pd.read_csv("health_dataset.csv")
df.head()

print("Hypothesis 1:")
print("H0: Smoking status has no effect on Diabetes prevalence.")
print("H1: Smoking status affects Diabetes prevalence.")

print("\nHypothesis 2:")
print("H0: Mean BMI of diabetic and non-diabetic patients is equal.")
print("H1: Mean BMI of diabetic and non-diabetic patients is different.")
```

## 2.Calculate Confidence Intervals

```python
from scipy import stats
age_mean = df['age'].mean()
age_sem = stats.sem(df['age'])
ci_age = stats.t.interval(0.95, len(df['age'])-1, loc=age_mean, scale=age_sem)
print(f"Confidence Interval of Age:{ci_age}")

weight_mean = df['weight'].mean()
weight_sem = stats.sem(df['weight'])
weight_age = stats.t.interval(0.95, len(df['weight'])-1, loc=weight_mean, scale=weight_sem)
print(f"Confidence Interval of Weight:{weight_age}")
```

## 3.Find the Critical Value and p-value && 4. T-test

```python
male_bmi = df[df['gender']=='Male']['bmi']
female_bmi = df[df['gender']=='Female']['bmi']

t_stat, p_value = stats.ttest_ind(
    male_bmi,
    female_bmi,
    equal_var= False
)

critical_value = stats.t.ppf(
    1-0.025,
    len(df)-2
)

print("T Statistic = ",t_stat)
print("P_value=",p_value)
print("Critical Value=",critical_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Accept H0")
```

## 5. Chi-square test

```python
contingency_table = pd.crosstab(
    df['smoking_status'],
    df['diabetes']
)

print("Contingency Table")
display(contingency_table)

chi2, p, dof, expected = stats.chi2_contingency(
    contingency_table
)

critical_chi = stats.chi2.ppf(
    0.95,
    dof
)

print("\nChi-Square Statistic:", chi2)
print("P Value:", p)
print("Degrees of Freedom:", dof)
print("Critical Value:", critical_chi)

if p < 0.05:
    print("\nDecision: Reject H0")
    print("Smoking and Diabetes are associated.")
else:
    print("\nDecision: Fail to Reject H0")
    print("No significant association found.")
```

## 6. ANOVA test

```python
# Create age groups

df['age_group'] = pd.cut(
    df['age'],
    bins=[0,30,45,60,100],
    labels=['Young','Middle','Senior','Old']
)

display(df[['age','age_group']].head())
```

```python
groups = []

for group_name, data in df.groupby('age_group'):
    groups.append(data['bmi'])

f_statistic, p_value = stats.f_oneway(*groups)

print("F Statistic:", f_statistic)
print("P Value:", p_value)

if p_value < 0.05:
    print("\nDecision: Reject H0")
    print("BMI differs across age groups.")
else:
    print("\nDecision: Fail to Reject H0")
    print("No significant difference.")
```

## 7. Covariance and Correlation

```python
covariance = np.cov(
    df['age'],
    df['bmi']
)[0,1]

correlation = df['age'].corr(
    df['bmi']
)

print("Covariance:", covariance)
print("Correlation:", correlation)
```
