# Part A – Theoretical Foundation (Short Notes)

## 1. What is Statistical Distribution?

A statistical distribution is a way of showing how data values are spread across different possible outcomes. It helps us understand the pattern, frequency, and probability of data values. Statistical distributions are widely used in statistics, data analysis, and machine learning to study data behavior.

## 2. What is a Q-Q Plot and Why is it Used?

A Q-Q (Quantile-Quantile) Plot is a graphical method used to compare a dataset with a theoretical distribution, usually the normal distribution. If the points on the graph form a straight line, the data follows the selected distribution. It is mainly used to check normality and identify outliers in data.

## 3. Difference Between Discrete and Continuous Distributions

A discrete distribution describes random variables that can take only specific countable values, such as the number of students in a class. A continuous distribution describes variables that can take any value within a range, such as height, weight, or temperature. Discrete data is counted, while continuous data is measured.

## 4. What is Bernoulli Distribution?

The Bernoulli Distribution is the simplest probability distribution and represents an experiment with only two possible outcomes: success (1) and failure (0). It is commonly used in situations such as coin tosses, where getting a head may be considered a success and a tail a failure.
### Probability Mass Function (PMF)

$$
P(X=x)=p^x(1-p)^{1-x}
$$

where,

$$
x \in \{0,1\}
$$

$$
0 \le p \le 1
$$

### Mean

$$
\mu = p
$$

### Variance

$$
\sigma^2 = p(1-p)
$$

## 5. What is Binomial Distribution?

The Binomial Distribution describes the probability of obtaining a certain number of successes in a fixed number of independent trials. Each trial has only two possible outcomes and the probability of success remains constant throughout the trials. For example, the number of heads obtained when tossing a coin ten times follows a binomial distribution.
### Probability Mass Function (PMF)

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$

where,

$$
\binom{n}{k}=\frac{n!}{k!(n-k)!}
$$

### Mean

$$
\mu = np
$$

### Variance

$$
\sigma^2 = np(1-p)
$$

## 6. Explain Log-Normal Distribution

A variable follows a Log-Normal Distribution when the logarithm of the variable is normally distributed. This distribution is always positive and usually appears right-skewed. It is commonly used to model stock prices, income levels, and biological growth processes.

## 7. Explain Power Law Distribution

A Power Law Distribution is a type of distribution where small values occur very frequently, while large values occur rarely. However, the rare large values can have a significant impact. Examples include city populations, wealth distribution, earthquake magnitudes, and social media followers.
### Equation

$$
P(X=x)=Cx^{-\alpha}
$$

where,

$$
C = \text{constant}
$$

$$
\alpha = \text{power-law exponent}
$$

## 8. What is Box-Cox Transform?

The Box-Cox Transform is a mathematical technique used to transform data into a more normal distribution. It helps reduce skewness, stabilize variance, and improve the performance of statistical and machine learning models. It is commonly used as a preprocessing step in data analysis.
### Transformation

$$
Y(\lambda)=\frac{Y^\lambda-1}{\lambda}, \quad \lambda \neq 0
$$

$$
Y(\lambda)=\ln(Y), \quad \lambda = 0
$$

## 9. Explain Poisson Distribution with an Example

The Poisson Distribution is used to model the number of times an event occurs within a fixed period of time or space. It assumes that events occur independently and at a constant average rate. For example, the number of calls received by a customer service center in one minute can be modeled using a Poisson distribution.
### Probability Mass Function (PMF)

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
$$

where,

$$
\lambda = \text{average rate of occurrence}
$$

### Mean

$$
\mu = \lambda
$$

### Variance

$$
\sigma^2 = \lambda
$$

## 10. What is Z-Score Probability?

A Z-score measures how far a data value is from the mean in terms of standard deviations. A positive Z-score indicates that the value is above the mean, while a negative Z-score indicates that it is below the mean. Z-scores are useful for calculating probabilities and identifying unusual observations in a dataset.
### Formula

$$
Z=\frac{X-\mu}{\sigma}
$$

where,

- $X$ = Observation
- $\mu$ = Mean
- $\sigma$ = Standard Deviation

### Standard Normal Distribution

$$
Z \sim N(0,1)
$$

## 11. Differentiate Probability Density Function (PDF) and Cumulative Distribution Function (CDF)

The Probability Density Function (PDF) describes how the probability of a continuous random variable is distributed over different values. It shows the relative likelihood of a variable taking a specific value. The Cumulative Distribution Function (CDF) gives the probability that a random variable is less than or equal to a particular value. While PDF shows probability density, CDF shows the accumulated probability up to a given point.
### Cumulative Distribution Function (CDF)

$$
F(x)=P(X \le x)
$$

### Relationship between PDF and CDF

$$
F(x)=\int_{-\infty}^{x} f(t)\,dt
$$

### PDF from CDF

$$
f(x)=\frac{d}{dx}F(x)
$$
