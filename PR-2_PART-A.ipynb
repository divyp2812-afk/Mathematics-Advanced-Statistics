# Part A – Theoretical Foundation

## 1.Inferential Statistics  
It uses sample data to make generalizations or predictions about a population, relying on probability theory.

## 2.Hypothesis Testing & Components  
A method to test assumptions about a population. Components: null hypothesis (H₀), alternative hypothesis (H₁), test statistic, significance level (α), and decision rule.

## 3.Confidence Interval & Critical Value  
A confidence interval gives a range of values likely to contain the true parameter. The critical value is the threshold beyond which we reject H₀.

## 4.p-value  
The probability of obtaining results at least as extreme as the observed, assuming H₀ is true. A small p-value indicates strong evidence against H₀.

## 5.Type I vs Type II Errors

Type I: Rejecting H₀ when it is true (false positive).

Type II: Failing to reject H₀ when H₁ is true (false negative).

## 6.Tests Overview

z-test: For large samples, known variance.

t-test: Small samples, unknown variance.

Chi-square test: Tests independence or goodness-of-fit for categorical data.

ANOVA: Compares means across 3+ groups.

## 7.Covariance  
Measures how two variables change together. Positive covariance means they move in the same direction, negative means opposite.

## 8.Correlation
Correlation indicates how two variables move together. A positive correlation means both increase/decrease together, while a negative correlation means one rises as the other falls. The correlation coefficient 𝑟 ranges from –1 to +1.
